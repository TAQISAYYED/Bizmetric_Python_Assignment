```python

# 37. Function to accept marks with exception validation
def accept_marks():
    try:
        marks = float(input("Enter marks (0-100): "))
        if marks < 0 or marks > 100:
            raise ValueError("Marks must be between 0 and 100")
        return marks
    except ValueError as e:
        print("Invalid Input:", e)


```

```python

# 41,42,43. Dictionary creation, save to file, read and append


import os

def add_customer():
    cust_dict = {}

    if os.path.exists("cust_info.txt"):
        with open("cust_info.txt", "r") as f:
            data = f.read()
            if data:
                cust_dict = eval(data)

    cust_no = len(cust_dict) + 1

    name = input("Enter Name: ")
    dob = input("Enter DOB (dd-mm-yyyy): ")
    mobile = input("Enter Mobile: ")

    cust_dict[cust_no] = {
        "name": name,
        "DOB": dob,
        "mobile": mobile
    }

    with open("cust_info.txt", "w") as f:
        f.write(str(cust_dict))

    print("Customer Added Successfully!")


```

```python
# 46. Function to calculate age till now
from datetime import datetime
def calculate_age(dob):
    dob_date = datetime.strptime(dob, "%d-%m-%Y")
    today = datetime.today()
    age = today.year - dob_date.year - ((today.month, today.day) < (dob_date.month, dob_date.day))
    return age


```

```python
# 47. Check age eligibility
def check_eligibility(dob, eligibility_age):
    age = calculate_age(dob)
    if age >= eligibility_age:
        return "Eligible"
    else:
        return "Not Eligible"
```

```python
# 48. Check palindrome
def check_palindrome(string):
    if string == string[::-1]:
        return "Palindrome"
    else:
        return "Not Palindrome"


```

```python
# 50. Factorial
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result
```

```python
# 52. Frequency of elements in list
def frequency(lst):
    freq = {}
    for i in lst:
        freq[i] = freq.get(i, 0) + 1
    return freq


```

```python

# 54. MySQL Tables + OOP Concept
"""
CREATE TABLE user(
    uid INT PRIMARY KEY AUTO_INCREMENT,
    mobile VARCHAR(10),
    passkey VARCHAR(20),
    date DATETIME
);

CREATE TABLE Product(
    pid INT PRIMARY KEY AUTO_INCREMENT,
    pname VARCHAR(50)
);

CREATE TABLE Orders(
    oid INT PRIMARY KEY AUTO_INCREMENT,
    uid INT,
    pid INT,
    FOREIGN KEY (uid) REFERENCES user(uid),
    FOREIGN KEY (pid) REFERENCES Product(pid)
);
"""

class User:

    def __init__(self, uid, mobile):
        self.uid = uid
        self.mobile = mobile

    def sign_up(self):
        print("User Signed Up")

    def sign_in(self):
        print("User Signed In")

    def view_products(self):
        print("Displaying Products")

    def view_orders(self):
        print("Displaying Orders")

    def place_order(self):
        print("Order Placed")


```

```python
# 56. Function to print bill
def print_bill(items):
    print("\n------ BILL ------")
    total = 0
    for item, price in items.items():
        print(f"{item} : {price}")
        total += price
    print("------------------")
    print("Total Amount:", total)
    print("------------------")
items = {
    "Pen": 10,
    "Notebook": 50,
    "Pencil": 5
}

print_bill(items)
```
